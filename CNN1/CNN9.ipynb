## Dataset Object

```python
def npy_loader(path):
    return torch.from_numpy(np.load(path))

class CTDataset(Dataset):
    def __init__(self, filepathImg, filepathLables):
        self.x = npy_loader(filepathImg).to(torch.float32)
        self.y = npy_loader(filepathLables)
        self.x = self.x / 255.
        self.y = torch.tensor(self.y, dtype=torch.long)
    def __len__(self): 
        return self.x.shape[0]
    def __getitem__(self, ix): 
        return self.x[ix], self.y[ix]
```

```python
ThinTrain_ds = CTDataset("BalancedDataBase\\ThinDataBase\\ThinTrainDataImages.npy","BalancedDataBase\\ThinDataBase\\ThinTrainDataLables.npy")
ThinTest_ds = CTDataset("BalancedDataBase\\ThinDataBase\\ThinTestDataImages.npy","BalancedDataBase\\ThinDataBase\\ThinTestDataLables.npy")
```

# A Rede

```python
class MyCNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv1 = nn.Conv2d(1,3,3,padding=1)
        self.pool = nn.MaxPool2d(2,2)
        self.conv2 = nn.Conv2d(3,9,3)
        self.fc1 = nn.Linear(9*6*6,25)
        self.fc2 = nn.Linear(25,25)
        self.fc3 = nn.Linear(25,6)
        self.R = nn.ReLU()
        self.dropout = nn.Dropout(p=0.25)
    def forward(self,x):
        x = self.pool(self.R(self.conv1(x)))
        x = self.dropout(x)
        x = self.pool(self.R(self.conv2(x)))
        x = self.dropout(x)
        x = x.view(-1,9*6*6)
        x = self.R(self.fc1(x))
        x = self.R(self.fc2(x))
        x = self.fc3(x)
        return x.squeeze()
```

```python
f = MyCNN()
```

## Treinamento

```python
def train_model(dl, f, n_epochs=20,LearnigRate=0.01):
    # Optimization
    opt = SGD(f.parameters(), lr=LearnigRate)
    L = nn.CrossEntropyLoss()

    # Train model
    losses = []
    epochs = []
    for epoch in range(n_epochs):
        print(f'Epoch {epoch}')
        N = len(dl)
        running_loss = 0
        for i, (x, y) in enumerate(dl):
            # Update the weights of the network
            opt.zero_grad() 
            loss_value = L(f(x.view(len(y),1,28,28)), y) 
            loss_value.backward() 
            opt.step() 
            # Store training data
            epochs.append(epoch+i/N)
            losses.append(loss_value.item())
            running_loss += loss_value.item()
        print(f'Loss: {running_loss / N:.4f}')
    return np.array(epochs), np.array(losses)
```

```python
Epocas = 30
LearningRate = 0.01
epoch_data, loss_data = train_model(ThinTrain_dl,f,n_epochs=Epocas,LearnigRate=LearningRate)
```

```python
plt.plot(epoch_data, loss_data)
plt.xlabel('Epoch Number')
plt.ylabel('Cross Entropy')
plt.title('Cross Entropy (per batch)')
```

```python
epoch_data_avgd = epoch_data.reshape(Epocas,-1).mean(axis=1)
loss_data_avgd = loss_data.reshape(Epocas,-1).mean(axis=1)
```

```python
plt.plot(epoch_data_avgd, loss_data_avgd, 'o--')
plt.xlabel('Epoch Number')
plt.ylabel('Cross Entropy')
plt.title('Cross Entropy (avgd per epoch)')
```
