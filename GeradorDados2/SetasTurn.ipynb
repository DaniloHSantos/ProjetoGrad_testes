```python
import cv2
import os
import matplotlib.pyplot as plt
```

```python
basepath = os.getcwd()

#Nomes base
SetaBaixoBoldBaseName = "SetaBaixo-Bold-"
SetaCimaBoldBaseName = "SetaCima-Bold-"
SetaDireitaBoldBaseName = "SetaDireita-Bold-"
SetaEsquerdaBoldBaseName = "SetaEsquerda-Bold-"

#Paths base
SetaBaixoPathBold = os.path.join(basepath,'DataBase\\Bold\\SetaBaixo')
SetaCimaPathBold = os.path.join(basepath,'DataBase\\Bold\\SetaCima')
SetaDireitaPathBold = os.path.join(basepath,'DataBase\\Bold\\SetaDireita')
SetaEsquerdaPathBold = os.path.join(basepath,'DataBase\\Bold\\SetaEsquerda')

#Save Paths
SetaBaixoSavePathBold = os.path.join(basepath,'NewSetas\\Bold\\SetaBaixo')
SetaCimaSavePathBold = os.path.join(basepath,'NewSetas\\Bold\\SetaCima')
SetaDireitaSavePathBold = os.path.join(basepath,'NewSetas\\Bold\\SetaDireita')
SetaEsquerdaSavePathBold = os.path.join(basepath,'NewSetas\\Bold\\SetaEsquerda')

#Seta Cima

#Percorre todas as imagens da pasta da Seta Cima
i=0
for filename in os.listdir(SetaCimaPathBold):
    print(filename)
    #Gera os números para os novos arquivos
    ImageNumber = i+851
    i += 1
    #Gera os nomes das novas imagens
    SetaBaixoBoldName = SetaBaixoBoldBaseName + str(ImageNumber) + ".jpg"
    SetaDireitaBoldName = SetaDireitaBoldBaseName + str(ImageNumber) + ".jpg"
    SetaEsquerdaBoldName = SetaEsquerdaBoldBaseName + str(ImageNumber) + ".jpg"
    
    #Carrega imagem
    imgCima = cv2.imread(os.path.join(SetaCimaPathBold,filename))
    imgBaixo = cv2.rotate(imgCima,cv2.ROTATE_180)
    imgDireita = cv2.rotate(imgCima,cv2.ROTATE_90_CLOCKWISE)
    imgEsquerda = cv2.rotate(imgCima,cv2.ROTATE_90_COUNTERCLOCKWISE)

    #Salva as imagens
    #Seta Baixo
    if not cv2.imwrite((os.path.join(SetaBaixoSavePathBold,SetaBaixoBoldName)), imgBaixo):
        raise Exception("Could not write the image")
    #Seta Direita
    if not cv2.imwrite((os.path.join(SetaDireitaSavePathBold,SetaDireitaBoldName)), imgDireita):
        raise Exception("Could not write the image")
    #Seta Esquerda
    if not cv2.imwrite((os.path.join(SetaEsquerdaSavePathBold,SetaEsquerdaBoldName)), imgEsquerda):
        raise Exception("Could not write the image")
print(i)
```
