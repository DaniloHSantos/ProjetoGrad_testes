```python
import numpy as np
import matplotlib.pyplot as plt

def generate_arrow_path(jitter=2, num_points=50):
    # Define key points of the arrow
    A = np.array([300, 100])  # Tip
    B = np.array([300, 500])  # Bottom of shaft
    C = np.array([250, 150])  # Left wing
    D = np.array([350, 150])  # Right wing

    def interpolate(p1, p2, steps):
        return [p1 + (p2 - p1) * t for t in np.linspace(0, 1, steps)]

    # Interpolate the 3 segments of the arrow
    shaft = interpolate(B, A, num_points)
    left_wing = interpolate(A, C, num_points // 2)
    right_wing = interpolate(A, D, num_points // 2)

    # Combine all parts into one path
    path = shaft + left_wing + right_wing

    # Add jitter
    jittered_path = [point + np.random.normal(0, jitter, 2) for point in path]

    return np.array(jittered_path)

# Example usage and plotting
arrow_path = generate_arrow_path()

plt.figure(figsize=(6, 6))
plt.plot(arrow_path[:, 0], arrow_path[:, 1], marker='o')
plt.title("Simulated Hand-Drawn Arrow")
plt.xlim(0, 600)
plt.ylim(600, 0)  # Invert y-axis for screen-like coordinates
plt.grid(True)
plt.gca().set_aspect('equal')
plt.show()
```
